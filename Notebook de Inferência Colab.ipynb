### Carregamento local e inferência

```python
# OBS.: LlamaIndex possui modelo de Embedding

# Carregamento do modelo

if True:
    from unsloth import FastLanguageModel
    model, tokenizer = FastLanguageModel.from_pretrained(
        model_name = "gustavomacedo/Especialista_MMGD_LORA",
        max_seq_length = 2048,
        dtype = None,
        load_in_4bit = True,
    )
    FastLanguageModel.for_inference(model) # classe que carrega o modelo com configurações assertivas

# Preparação de prompt

alpaca_prompt = """Abaixo está uma Pergunta. Escreva uma Resposta que responda apropriadamente a Pergunta. {}
### Pergunta:
{}

### Resposta:
{}"""
```

```python
# Inferência

inputs = tokenizer(
[
    alpaca_prompt.format(
        "", # coloque aqui o seu prompt
        "Onde encontro as normas sobre MMGD no Brasil?", # input
        "", # deixe em branco, o modelo vai preencher!
    )
], return_tensors = "pt").to("cuda")

from transformers import TextStreamer
text_streamer = TextStreamer(tokenizer)
_ = model.generate(**inputs, streamer = text_streamer, max_new_tokens = 700)
```
